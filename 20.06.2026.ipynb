```python
import nltk
# Download required NLTK resources
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('vader_lexicon')

from nltk.classify import NaiveBayesClassifier
from nltk.tokenize import word_tokenize
from nltk import FreqDist, bigrams, trigrams
from nltk.probability import MLEProbDist
from nltk.sem import Expression

# Define the features function (assuming it extracts word features)
def features(text):
    words = word_tokenize(text.lower())
    return {word: True for word in words if word.isalpha()}

# Define the news text (assuming this variable was defined earlier)
news = "AI technology is improving business operations through cloud computing platforms"

# Training data
train_data = [
    (features("artificial intelligence machine learning programming cloud platform"), "Technology"),
    (features("cricket football sports tournament"), "Sports"),
    (features("government election parliament politics"), "Politics"),
    (features("stock market finance business company"), "Business")
]

classifier = NaiveBayesClassifier.train(train_data)

category = classifier.classify(features(news))
print("\nCategory:")
print(category)

# --------------------------------------------------
# 4. Model Evaluation
# --------------------------------------------------
test_data = [
    (features("AI platform technology"), "Technology"),
    (features("football match"), "Sports")
]

correct = 0
for feat, label in test_data:
    if classifier.classify(feat) == label:
        correct += 1

accuracy = correct / len(test_data)

print("\nModel Accuracy:")
print(f"{accuracy*100:.0f}%")

# --------------------------------------------------
# 5. Word Frequency & Probability Distribution
# --------------------------------------------------
words = [w.lower() for w in word_tokenize(news)
         if w.isalpha()]

freq_dist = FreqDist(words)

print("\nWord Frequency:")
for word, freq in freq_dist.items():
    print(word, ":", freq)

prob_dist = MLEProbDist(freq_dist)

print("\nProbability Distribution:")
for word in ["ai", "technology", "business"]:
    if word in freq_dist:
        print(f"P({word}) = {prob_dist.prob(word):.2f}")

# --------------------------------------------------
# 6. Semantic Analysis using nltk.sem
# --------------------------------------------------
print("\nSemantic Representation:")
read_expr = Expression.fromstring
expr = read_expr('improve(technology,business)')
print(expr)

# --------------------------------------------------
# 7. Bigrams and Trigrams
# --------------------------------------------------
token_words = [w for w in word_tokenize(news) if w.isalpha()]

print("\nBigrams:")
for bg in list(bigrams(token_words)):
    print(bg)

print("\nTrigrams:")
for tg in list(trigrams(token_words)):
    print(tg)
```
