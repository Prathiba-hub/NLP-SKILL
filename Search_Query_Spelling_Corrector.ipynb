```python
import re
import pandas as pd
import numpy as np
```

```python
corpus_file = "birkbeck.txt"

with open(corpus_file, "r", encoding="utf-8") as file:
    corpus_text = file.read()

print("Birkbeck corpus loaded successfully!")
print("Number of characters:", len(corpus_text))
```

```python
words = re.findall(r"[a-zA-Z]+", corpus_text.lower())

print("Total words found:", len(words))
print("\nFirst 50 words:")
print(words[:50])
```

```python
vocabulary = set(words)

print("Vocabulary size:", len(vocabulary))
```

```python
def tokenize_query(query):
    query = query.lower()
    tokens = re.findall(r"[a-zA-Z]+", query)
    return tokens
```

```python
query = "machine lerning cours"

tokens = tokenize_query(query)

print("Original query:")
print(query)

print("\nTokens:")
print(tokens)
```

```python
def find_misspelled_words(tokens, vocabulary):
    
    misspelled = []
    
    for word in tokens:
        if word not in vocabulary:
            misspelled.append(word)
    
    return misspelled
```

```python
def edit_distance(word1, word2):
    
    m = len(word1)
    n = len(word2)
    
    # Create distance matrix
    dp = np.zeros((m + 1, n + 1), dtype=int)
    
    # Initialize first column
    for i in range(m + 1):
        dp[i][0] = i
    
    # Initialize first row
    for j in range(n + 1):
        dp[0][j] = j
    
    # Calculate distances
    for i in range(1, m + 1):
        
        for j in range(1, n + 1):
            
            if word1[i - 1] == word2[j - 1]:
                cost = 0
            else:
                cost = 1
            
            dp[i][j] = min(
                dp[i - 1][j] + 1,       # Deletion
                dp[i][j - 1] + 1,       # Insertion
                dp[i - 1][j - 1] + cost # Substitution
            )
    
    return dp[m][n]
```

```python
def suggest_correction(word, vocabulary):
    
    best_word = None
    best_distance = float("inf")
    
    for candidate in vocabulary:
        
        distance = edit_distance(word, candidate)
        
        if distance < best_distance:
            best_distance = distance
            best_word = candidate
    
    return best_word, best_distance
```

```python
word = "lerning"

suggestion, distance = suggest_correction(
    word,
    vocabulary
)

print("Incorrect word:", word)
print("Suggested word:", suggestion)
print("Edit distance:", distance)
```

```python
def get_candidates(word, vocabulary, max_length_difference=2):
    
    candidates = [
        candidate
        for candidate in vocabulary
        if abs(len(candidate) - len(word)) <= max_length_difference
    ]
    
    return candidates
```

```python
def suggest_correction(word, vocabulary):
    
    candidates = get_candidates(
        word,
        vocabulary
    )
    
    # If no candidates are found,
    # use the complete vocabulary
    if len(candidates) == 0:
        candidates = vocabulary
    
    best_word = None
    best_distance = float("inf")
    
    for candidate in candidates:
        
        distance = edit_distance(
            word,
            candidate
        )
        
        if distance < best_distance:
            best_distance = distance
            best_word = candidate
    
    return best_word, best_distance
```

```python
def correct_query(query, vocabulary):
    
    tokens = tokenize_query(query)
    
    corrected_tokens = []
    corrections = []
    
    for word in tokens:
        
        # If word is already correct
        if word in vocabulary:
            corrected_tokens.append(word)
        
        # If word is misspelled
        else:
            suggestion, distance = suggest_correction(
                word,
                vocabulary
            )
            
            corrected_tokens.append(suggestion)
            
            corrections.append({
                "incorrect_word": word,
                "suggested_word": suggestion,
                "edit_distance": distance
            })
    
    corrected_query = " ".join(corrected_tokens)
    
    return corrected_query, corrections
```

```python
query = "machine lerning cours"

corrected_query, corrections = correct_query(
    query,
    vocabulary
)

print("Original Query:")
print(query)

print("\nCorrections:")

for correction in corrections:
    print(
        correction["incorrect_word"],
        "→",
        correction["suggested_word"],
        "(Distance:",
        correction["edit_distance"],
        ")"
    )

print("\nCorrected Query:")
print(corrected_query)
```

```python
corrections_df = pd.DataFrame(corrections)

corrections_df
```

```python
example_queries = [
    "machine lerning cours",
    "computr science",
    "artifical inteligence",
    "data anlysis",
    "softwere development"
]

results = []

for query in example_queries:
    
    corrected_query, corrections = correct_query(
        query,
        vocabulary
    )
    
    results.append({
        "Original Query": query,
        "Corrected Query": corrected_query,
        "Number of Errors": len(corrections)
    })

results_df = pd.DataFrame(results)

results_df
```
