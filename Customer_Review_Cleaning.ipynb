```python
import re
import string
import pandas as pd
import nltk

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")
```

```python
df = pd.read_csv("IMDB Dataset.csv")

print("Dataset loaded successfully!")
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
df.head()
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
duplicate_count = df["review"].duplicated().sum()

print("Number of duplicate reviews:", duplicate_count)
```

```python
df = df.drop_duplicates(subset=["review"]).copy()

print("Rows after removing duplicate reviews:", len(df))
```

```python
stop_words = set(stopwords.words("english"))

lemmatizer = WordNetLemmatizer()

print("Number of stopwords:", len(stop_words))
```

```python
def clean_review(text):
    
    # Convert to string
    text = str(text)
    
    # 1. Convert to lowercase
    text = text.lower()
    
    # 2. Remove HTML tags
    text = re.sub(r"<.*?>", " ", text)
    
    # 3. Remove URLs
    text = re.sub(r"http\S+|www\S+|https\S+", " ", text)
    
    # 4. Remove numbers
    text = re.sub(r"\d+", " ", text)
    
    # 5. Remove punctuation and special characters
    text = re.sub(r"[^a-zA-Z\s]", " ", text)
    
    # 6. Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()
    
    # 7. Tokenize
    words = word_tokenize(text)
    
    # 8. Remove stopwords
    words = [
        word for word in words
        if word not in stop_words
    ]
    
    # 9. Lemmatization
    words = [
        lemmatizer.lemmatize(word)
        for word in words
    ]
    
    # 10. Join words back into a sentence
    cleaned_text = " ".join(words)
    
    return cleaned_text
```

```python
sample_text = 'This movie was AMAZING!!! 😊'

print("Original:")
print(sample_text)

print("\nCleaned:")
print(clean_review(sample_text))
```

```python
sample_text = """
This movie was AMAZING!!! 😊
<br>
Visit https://example.com for more information.
I watched it 10 times!!!
"""

print("Original:")
print(sample_text)

print("\nCleaned:")
print(clean_review(sample_text))
```

```python
print("Cleaning reviews...")
print("Please wait...")

df["cleaned_review"] = df["review"].apply(clean_review)

print("Cleaning completed!")
```
