```python
%pip install datasets nltk numpy -q

import re
import nltk
from collections import Counter, defaultdict
from datasets import load_dataset

nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)

dataset = load_dataset("Salesforce/wikitext", "wikitext-2-raw-v1")

texts = dataset["train"]["text"]

tokens = []

for text in texts:
    text = text.lower()
    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)
    text = re.sub(r"\s+", " ", text).strip()
    
    if text:
        tokens.extend(nltk.word_tokenize(text))

bigram = Counter(zip(tokens[:-1], tokens[1:]))
trigram = Counter(zip(tokens[:-2], tokens[1:-1], tokens[2:]))

bigram_next = defaultdict(Counter)
trigram_next = defaultdict(Counter)

for (w1, w2), count in bigram.items():
    bigram_next[w1][w2] = count

for (w1, w2, w3), count in trigram.items():
    trigram_next[(w1, w2)][w3] = count

def predict_next_words(sentence, top_n=5):
    words = re.findall(r"[a-zA-Z0-9]+", sentence.lower())

    if not words:
        return []

    candidates = {}

    if len(words) >= 2:
        candidates = trigram_next.get((words[-2], words[-1]), {})

    if not candidates:
        candidates = bigram_next.get(words[-1], {})

    if not candidates:
        return []

    total = sum(candidates.values())

    predictions = [
        (word, count / total)
        for word, count in candidates.items()
    ]

    predictions.sort(key=lambda x: x[1], reverse=True)

    return predictions[:top_n]

print("WikiText-2 loaded successfully.")
print("Next-Word Predictor is ready.")

query = input("\nEnter a sentence or partial sentence: ")

predictions = predict_next_words(query, 5)

print("\nInput:", query)
print("\nTop predicted next words:")

if predictions:
    for i, (word, probability) in enumerate(predictions, 1):
        print(f"{i}. {word} - {probability:.3f}")
else:
    print("No prediction found.")
```
